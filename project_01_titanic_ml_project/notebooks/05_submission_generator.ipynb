```python
# 05_submission_generator

import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# 전처리된 학습 데이터 불러오기
df = pd.read_csv('../outputs/clean_train.csv')
X = df.drop('Survived', axis=1)
y = df['Survived']

# 모델 학습
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)

# ✅ 원본 test.csv에서 PassengerId 따로 추출
test_raw = pd.read_csv('../data/test.csv')
passenger_ids = test_raw['PassengerId']

# 전처리된 테스트셋 로딩
test_clean = pd.read_csv('../outputs/clean_test.csv')

# 예측
y_pred = model.predict(test_clean)

# 결과 저장
submission = pd.DataFrame({
    'PassengerId': passenger_ids,
    'Survived': y_pred
})
submission.to_csv('../outputs/submission.csv', index=False)
print("🚀 submission.csv 저장 완료!")
```
