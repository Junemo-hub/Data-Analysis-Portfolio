```python
import pandas as pd
import matplotlib.pyplot as plt

# 이미 수집한 결과 딕셔너리
metrics = {
    "Naive Bayes": {"recall_0": 0.27, "f1_0": 0.42, "roc_auc": 0.937},
    "Logistic": {"recall_0": 0.67, "f1_0": 0.75, "roc_auc": 0.957},
    "Logistic + Balanced": {"recall_0": 0.81, "f1_0": 0.80, "roc_auc": 0.954},
    "Random Forest + Balanced": {"recall_0": 0.78, "f1_0": 0.77, "roc_auc": 0.955}
}

df_metrics = pd.DataFrame(metrics).T
df_metrics.columns = ["Recall (0)", "F1 (0)", "ROC AUC"]
df_metrics = df_metrics.round(3)
df_metrics

# 값 확인 필요
```

```python
df_metrics.plot(kind='bar', figsize=(10, 6))
plt.title("Model Performance on Negative Class")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=15)
plt.grid(axis='y')
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()
```

```python
## 🧠 Final Comparison Summary

#- **Logistic Regression (balanced)** achieved the highest Recall and F1-score on negative sentiment, making it the most effective in class imbalance.
#- **Naive Bayes** was fast but struggled to detect negative samples.
#- **Random Forest (balanced)** performed well overall but took significantly longer to train.
#- **Conclusion**: Logistic + Balanced is the recommended choice for balanced performance and interpretability.
```
