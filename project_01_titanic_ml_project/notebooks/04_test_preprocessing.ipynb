```python
import pandas as pd
import os

def preprocess(df):
    df = df.copy()
    df['Age'] = df['Age'].fillna(df['Age'].median())
    df['Fare'] = df['Fare'].fillna(df['Fare'].median())
    df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
    df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
    df = pd.get_dummies(df, columns=['Embarked'], drop_first=True)
    df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
    df['IsAlone'] = (df['FamilySize'] == 1).astype(int)
    return df

# 원본 로딩
test_df_raw = pd.read_csv('../data/test.csv')
passenger_ids = test_df_raw['PassengerId']

# 전처리
test_df = preprocess(test_df_raw)

# 불필요한 열 제거
test_df = test_df.drop(['PassengerId', 'Name', 'Ticket', 'Cabin'], axis=1, errors='ignore')

# 저장
os.makedirs('../outputs', exist_ok=True)
test_df.to_csv('../outputs/clean_test.csv', index=False)
print("✅ clean_test.csv 저장 완료")
```
