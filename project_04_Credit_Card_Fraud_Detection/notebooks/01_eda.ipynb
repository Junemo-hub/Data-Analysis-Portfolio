```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set(style="whitegrid")  # 보기 좋은 테마
```

```python
df['Class'].value_counts()
```

```python

df['Class'].value_counts(normalize=True)
```

```python
sns.countplot(x='Class', data=df)
```

```python
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.histplot(df['Amount'], bins=100)
plt.title("Original Amount")

plt.subplot(1, 2, 2)
sns.histplot(np.log1p(df['Amount']), bins=100)
plt.title("Log1p Transformed Amount")
plt.show()

```

```python
df[df['Class'] == 1]['Amount'].describe()
df[df['Class'] == 0]['Amount'].describe()

sns.boxplot(x='Class', y='Amount', data=df)
```

```python
corr = df.corr()
plt.figure(figsize=(12,10))
sns.heatmap(corr, cmap='coolwarm', center=0)
```

```python
print(f"Fraud Ratio: {df['Class'].mean():.5f}")
```

```python
for col in ['V1', 'V2', 'V3', 'V4', 'V10']:
    sns.kdeplot(df[df['Class'] == 0][col], label='Normal', fill=True)
    sns.kdeplot(df[df['Class'] == 1][col], label='Fraud', fill=True)
    plt.title(f"Distribution for {col}")
    plt.show()

# 예시고 전체 V에 대해서 해도 좋음.
```

```python
df['Hour'] = (df['Time'] // 3600) % 24
sns.countplot(x='Hour', hue='Class', data=df)
```
