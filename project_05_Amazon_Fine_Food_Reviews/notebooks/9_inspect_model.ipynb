```python
import joblib

# 모델 불러오기
model = joblib.load('../outputs/models/logistic_balanced.pkl')

# 모델 구조 보기
print("모델 클래스:", type(model))
print("하이퍼파라미터:", model.get_params())
```

```python
vec = joblib.load('../outputs/vectorizer_logistic.pkl')

# 단어 20개 출력
features = vec.get_feature_names_out()
print("벡터라이저 feature 개수:", len(features))
print("샘플 단어:", features[:20])
```

```python
import numpy as np
coefs = model.coef_[0]
top_pos = np.argsort(coefs)[-10:]
top_neg = np.argsort(coefs)[:10]

print("Top Positive Words:", features[top_pos])
print("Top Negative Words:", features[top_neg])
```
